```python

import os
import shutil
from pathlib import Path

def copy_data(src_img_dir, src_mask_dir, dst_img_dir, dst_mask_dir, frac = 1.0):
    Path(dst_img_dir).mkdir(parents=True, exist_ok=True)
    Path(dst_mask_dir).mkdir(parents=True, exist_ok=True)
    
    images = os.listdir(src_img_dir)
    images = images[:int(frac * len(images))]
    for img_name in os.listdir(src_img_dir):
        img_path = os.path.join(src_img_dir, img_name)
        mask_path = os.path.join(src_mask_dir, img_name)

        if os.path.isfile(img_path) and os.path.exists(mask_path):
            shutil.copy2(img_path, os.path.join(dst_img_dir, img_name))
            shutil.copy2(mask_path, os.path.join(dst_mask_dir, img_name))

src_img_dir = "/mnt/data/omkumar/foundation_model/shared_foundation/datasets/US_Nerve/images"
src_mask_dir = "/mnt/data/omkumar/foundation_model/shared_foundation/datasets/US_Nerve/masks"

dest_img_dir = "/mnt/data/omkumar/foundation_model/shared_foundation/datasets/US_Nerve_10/images"
dest_mask_dir = "/mnt/data/omkumar/foundation_model/shared_foundation/datasets/US_Nerve_10/masks"

copy_data(src_img_dir,src_mask_dir,dest_img_dir,dest_mask_dir,0.1)
```
