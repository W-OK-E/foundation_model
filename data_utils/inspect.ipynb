```python
import tqdm
import os
import cv2
masks_dir = '/mnt/data/omkumar/foundation_phase1/datasets/cholec_1fps/masks/'
masks = os.listdir(masks_dir)

#check if all the images have the same shape or not if yes, reshape to 480x854 from 299x299
global_shape = [480, 854]
for mask_name in tqdm.tqdm(masks):
    mask_path = os.path.join(masks_dir, mask_name)
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
    mask_shape = mask.shape
    if global_shape is None:
        global_shape = mask.shape
        print("Got Global Shape:", global_shape)
    else:
        if mask_shape[0] != global_shape[0] or mask_shape[1] != global_shape[1]:
            mask = cv2.resize(mask, (global_shape[1], global_shape[0]))
            print(f"Resized mask {mask_name} from {mask.shape} to {global_shape}")
            print(f"Saving mask{mask_path}")
            cv2.imwrite(mask_path, mask)
```

```python
img_path = '/mnt/data/omkumar/foundation_phase1/datasets/cholec_1fps/images'
mask_path = '/mnt/data/omkumar/foundation_phase1/datasets/cholec_1fps/masks'

#snippet to check the which images are not rgb and to remove them and the corresponding masks
import os
from PIL import Image
import tqdm

count = 0
for filename in tqdm.tqdm(os.listdir(img_path)):
    image_file = os.path.join(img_path, filename)
    mask_file = os.path.join(mask_path, filename)

    try:
        with Image.open(image_file) as img:
            if img.mode != 'RGB':
                print(f"Removing {filename}: mode is {img.mode}")
                os.remove(image_file)
                if os.path.exists(mask_file):
                    os.remove(mask_file)
                count += 1
    except Exception as e:
        print(f"Error processing {filename}: {e}")

print("Images that will be removed:",count)
```

```python
import tqdm 
import cv2
import numpy as np
import os
import torch

m1_path = '/mnt/data/omkumar/mounted_datasets/Mammogrpahy/INBreast/extras/CalcificationSegmentationMasks'
m2_path = '/mnt/data/omkumar/mounted_datasets/Mammogrpahy/INBreast/extras/MassSegmentationMasks'
final_mask_path = '/mnt/data/omkumar/foundation_phase1/datasets/INBreast/INBreast/masks_pt'

os.makedirs(final_mask_path,exist_ok = True)

for m1 in tqdm.tqdm(os.listdir(m1_path)):
    m1_mask = cv2.imread(os.path.join(m1_path,m1),cv2.IMREAD_GRAYSCALE)
    m2_mask = cv2.imread(os.path.join(m2_path,m1),cv2.IMREAD_GRAYSCALE)

    save_path = os.path.join(final_mask_path,m1.replace("png","pt")) 
    m1_mask_np = np.zeros_like(m1_mask)
    m2_mask_np = np.zeros_like(m1_mask)

    m1_mask_np[m1_mask > 0] = 1
    if m2_mask is not None:
        m2_mask_np[m2_mask>0] = 1

    joint_mask = np.stack([m1_mask_np,m2_mask_np],axis = -1)
    joint_mask_tensor = torch.from_numpy(joint_mask)

    # Optional: ensure desired dtype
    # joint_mask_tensor = joint_mask_tensor.float()

    torch.save(joint_mask_tensor, save_path)
    # print("Joint mask shape:",joint_mask.shape)
    
```
