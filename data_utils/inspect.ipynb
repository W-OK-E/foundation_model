```python
import tqdm 
import cv2
import numpy as np
import os
import torch

m1_path = '/mnt/data/omkumar/mounted_datasets/Mammogrpahy/INBreast/extras/CalcificationSegmentationMasks'
m2_path = '/mnt/data/omkumar/mounted_datasets/Mammogrpahy/INBreast/extras/MassSegmentationMasks'
final_mask_path = '/mnt/data/omkumar/foundation_phase1/datasets/INBreast/INBreast/masks_pt'

os.makedirs(final_mask_path,exist_ok = True)

for m1 in tqdm.tqdm(os.listdir(m1_path)):
    m1_mask = cv2.imread(os.path.join(m1_path,m1),cv2.IMREAD_GRAYSCALE)
    m2_mask = cv2.imread(os.path.join(m2_path,m1),cv2.IMREAD_GRAYSCALE)

    save_path = os.path.join(final_mask_path,m1.replace("png","pt")) 
    m1_mask_np = np.zeros_like(m1_mask)
    m2_mask_np = np.zeros_like(m1_mask)

    m1_mask_np[m1_mask > 0] = 1
    if m2_mask is not None:
        m2_mask_np[m2_mask>0] = 1

    joint_mask = np.stack([m1_mask_np,m2_mask_np],axis = -1)
    joint_mask_tensor = torch.from_numpy(joint_mask)

    # Optional: ensure desired dtype
    # joint_mask_tensor = joint_mask_tensor.float()

    torch.save(joint_mask_tensor, save_path)
    # print("Joint mask shape:",joint_mask.shape)
    
```
