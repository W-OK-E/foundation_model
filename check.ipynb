```python
import h5py
import numpy as np
import glob

def load_full_volume(volume_num, base_path):
    slice_files = sorted(glob.glob(f"{base_path}/volume_{volume_num}_slice_*.h5"))
    slices = []
    for sf in slice_files:
        with h5py.File(sf, 'r') as f:
            slice_data = f['data'][:]  # Adjust key if needed
            slices.append(slice_data)
    volume = np.stack(slices, axis=0)  # Stack along slice axis
    return volume
```
